**Description**

Prognostics and health management is an important topic in industry for predicting state of assets to avoid downtime and failures. This data set is the Kaggle version of the very well known public data set for asset degradation modeling from NASA. It includes Run-to-Failure simulated data from turbo fan jet engines.

Engine degradation simulation was carried out using C-MAPSS. Four different were sets simulated under different combinations of operational conditions and fault modes. Records several sensor channels to characterize fault evolution. The data set was provided by the Prognostics CoE at NASA Ames.

<br>

**Prediction Goal**

In this dataset the goal is to predict the remaining useful life (RUL) of each engine in the test dataset. RUL is equivalent of number of flights remained for the engine after the last datapoint in the test dataset.

<br>

**Experimental Scenario**

Data sets consists of multiple multivariate time series. Each data set is further divided into training and test subsets. Each time series is from a different engine i.e., the data can be considered to be from a fleet of engines of the same type. Each engine starts with different degrees of initial wear and manufacturing variation which is unknown to the user. This wear and variation is considered normal, i.e., it is not considered a fault condition. There are three operational settings that have a substantial effect on engine performance. These settings are also included in the data. The data is contaminated with sensor noise.

The engine is operating normally at the start of each time series, and develops a fault at some point during the series. In the training set, the fault grows in magnitude until system failure. In the test set, the time series ends some time prior to system failure. The objective of the competition is to predict the number of remaining operational cycles before failure in the test set, i.e., the number of operational cycles after the last cycle that the engine will continue to operate. Also provided a vector of true Remaining Useful Life (RUL) values for the test data.

The data are provided as a zip-compressed text file with 26 columns of numbers, separated by spaces. Each row is a snapshot of data taken during a single operational cycle, each column is a different variable. The columns correspond to:

<br>

1) unit number

2) time, in cycles

3) operational setting 1

4) operational setting 2

5) operational setting 3

6) sensor measurement 1

7) sensor measurement 2

…

26) sensor measurement 26

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

# ==========================================
# 1. Data Preparation & Splitting
# ==========================================
# Assuming df_train and df_test are loaded
# Check if settings are the first 3 columns (Standard in CMAPSS)
n_settings = 3

X_train_raw = df_train.iloc[:, 2:-1].values
y_train = df_train.iloc[:, -1].values
X_test_raw = df_test.iloc[:, 2:-1].values
y_test = df_test.iloc[:, -1].values

# Scaling (Scale first, then split to keep distribution consistent)
scaler = StandardScaler()
X_train_all = scaler.fit_transform(X_train_raw)
X_test_all = scaler.transform(X_test_raw)

# Split into Controls (Settings) and Sensors (Measurements)
# Train Set
X_train_ctrl = X_train_all[:, :n_settings]      # First 3 columns
X_train_sens = X_train_all[:, n_settings:]      # Remaining columns
# Test Set
X_test_ctrl = X_test_all[:, :n_settings]
X_test_sens = X_test_all[:, n_settings:]

# Convert to Tensors
X_train_ctrl_t = torch.FloatTensor(X_train_ctrl)
X_train_sens_t = torch.FloatTensor(X_train_sens)
y_train_t = torch.FloatTensor(y_train).unsqueeze(1)

X_test_ctrl_t = torch.FloatTensor(X_test_ctrl)
X_test_sens_t = torch.FloatTensor(X_test_sens)
y_test_t = torch.FloatTensor(y_test).unsqueeze(1)

# ==========================================
# 2. Define Proposed Model (Serial Encoder)
# ==========================================
class SerialRULModel(nn.Module):
    def __init__(self, n_ctrl, n_sens, n_latent=4):
        super(SerialRULModel, self).__init__()
        
        # Encoder 1: Compresses Control Inputs (Settings) -> h
        self.control_encoder = nn.Sequential(
            nn.Linear(n_ctrl, 8),
            nn.ReLU(),
            nn.Linear(8, n_latent), # Compress to vector h
            nn.ReLU()
        )
        
        # Encoder 2: Takes (Sensors + h) -> Prediction
        self.main_predictor = nn.Sequential(
            nn.Linear(n_sens + n_latent, 64), # Concatenation happens here
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )
        
    def forward(self, x_ctrl, x_sens):
        # 1. Get latent vector h from controls
        h = self.control_encoder(x_ctrl)
        
        # 2. Concatenate h with sensor data
        # dim=1 means stack them side-by-side
        combined = torch.cat((x_sens, h), dim=1)
        
        # 3. Final Prediction
        out = self.main_predictor(combined)
        return out

# Initialize Model
model = SerialRULModel(n_ctrl=X_train_ctrl.shape[1], 
                       n_sens=X_train_sens.shape[1], 
                       n_latent=5) # Latent vector size (adjustable)

optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.MSELoss()

# ==========================================
# 3. Training Loop
# ==========================================
epochs = 3000
loss_history = []

print(f"Total Epochs : {epochs}")
print("Training Proposed Serial Model...")

for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    
    # Pass both inputs
    outputs = model(X_train_ctrl_t, X_train_sens_t)
    loss = criterion(outputs, y_train_t)
    
    loss.backward()
    optimizer.step()
    
    # Evaluation
    model.eval()
    with torch.no_grad():
        test_pred = model(X_test_ctrl_t, X_test_sens_t)
        test_loss = criterion(test_pred, y_test_t).item()
        loss_history.append(test_loss)
    
    if (epoch+1) % 300 == 0:
         print(f"At Epoch : {epoch+1} -> Test_MSE: {test_loss:.4f}")

# ==========================================
# 4. Compare with Linear Baseline
# ==========================================
# Recalculate Linear Baseline for comparison
X_train_b = np.hstack((np.ones((X_train_all.shape[0],1)), X_train_all))
X_test_b = np.hstack((np.ones((X_test_all.shape[0],1)), X_test_all))
theta_opt = np.linalg.lstsq(X_train_b, y_train, rcond=None)[0]
y_pred_lin = X_test_b.dot(theta_opt)
mse_linear = np.mean((y_pred_lin - y_test)**2)

# ==========================================
# 5. Visualization
# ==========================================
plt.style.use('default') # Force white background
plt.figure(figsize=(12, 6))

# Plot Proposed Model
plt.plot(loss_history, label='Proposed Serial Model (Test MSE)', color='blue', linewidth=2)

# Plot Linear Baseline
plt.axhline(y=mse_linear, color='red', linestyle='--', linewidth=2, label=f'Linear Baseline (MSE={mse_linear:.2f})')

plt.title('Performance: Proposed Serial Architecture vs Baseline', fontsize=14)
plt.xlabel('Epochs', fontsize=12)
plt.ylabel('Test MSE', fontsize=12)
plt.legend(facecolor='white', framealpha=1, edgecolor='black')
plt.grid(True, linestyle=':', alpha=0.6)
plt.ylim(0, mse_linear * 1.5) # Zoom in

plt.show()

print("\n[Final Result Comparison]")
print(f"Linear Baseline MSE: {mse_linear:.4f}")
print(f"Proposed Model MSE:  {loss_history[-1]:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

X_train_raw=df_train.iloc[:,2:-1].values
y_train=df_train.iloc[:,-1].values
X_test_raw=df_test.iloc[:,2:-1].values
y_test=df_test.iloc[:,-1].values

scaler=StandardScaler()
X_train=scaler.fit_transform(X_train_raw)
X_test=scaler.transform(X_test_raw)

# Module
lin_reg=LinearRegression()
lin_reg.fit(X_train,y_train)

y_train_pred_mod=lin_reg.predict(X_train)
train_mse_mod=np.mean((y_train_pred_mod-y_train)**2)

y_pred_mod=lin_reg.predict(X_test)
test_mse_mod=np.mean((y_pred_mod-y_test)**2)

# Formula
X_train_b=np.hstack((np.ones((X_train.shape[0],1)),X_train))
X_test_b=np.hstack((np.ones((X_test.shape[0],1)),X_test))

theta_opt=np.linalg.lstsq(X_train_b,y_train,rcond=None)[0]

y_train_pred_anal=X_train_b.dot(theta_opt)
train_mse_anal=np.mean((y_train_pred_anal-y_train)**2)

y_pred_anal=X_test_b.dot(theta_opt)
test_mse_anal=np.mean((y_pred_anal-y_test)**2)

# Deep_Learning
X_train_t=torch.FloatTensor(X_train)
y_train_t=torch.FloatTensor(y_train).unsqueeze(1)
X_test_t=torch.FloatTensor(X_test)
y_test_t=torch.FloatTensor(y_test).unsqueeze(1)

class RULModel(nn.Module):
    def __init__(self,input_dim):
        super(RULModel,self).__init__()
        self.net=nn.Sequential(
            nn.Linear(input_dim,64),
            nn.ReLU(),
            nn.Linear(64,32),
            nn.ReLU(),
            nn.Linear(32,1)
        )
    def forward(self,x):
        return self.net(x)

model=RULModel(X_train.shape[1])
optimizer=optim.Adam(model.parameters(),lr=0.01)
criterion=nn.MSELoss()

epochs=3000
dl_mse_history=[]
interval=epochs//10
print(f"Total Epochs : {epochs}")

for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()
    outputs=model(X_train_t)
    loss=criterion(outputs,y_train_t)
    loss.backward()
    optimizer.step()
    
    model.eval()
    with torch.no_grad():
        test_pred=model(X_test_t)
        test_loss=criterion(test_pred,y_test_t).item()
        dl_mse_history.append(test_loss)
        
    if(epoch+1)%interval==0:
        print(f"At Epoch : {epoch+1} -> Train_MSE: {loss.item():.4f} ,Test_MSE: {test_loss:.4f}")

model.eval()
with torch.no_grad():
    final_train_pred=model(X_train_t)
    final_train_mse=criterion(final_train_pred,y_train_t).item()
    final_test_pred=model(X_test_t)
    final_test_mse=criterion(final_test_pred,y_test_t).item()

plt.figure(figsize=(12,6))
plt.axhline(y=test_mse_mod,color='green',linestyle='--',linewidth=2,label=f'Linear Module Test MSE ({test_mse_mod:.4f})')
plt.axhline(y=test_mse_anal,color='red',linestyle=':',linewidth=3,alpha=0.8,label=f'Linear Analytical Test MSE ({test_mse_anal:.4f})')
plt.plot(dl_mse_history,label='Deep Learning Test MSE',color='blue',linewidth=2)
plt.title('Performance Comparison: Module vs Formula vs Deep Learning',fontsize=14)
plt.xlabel('Epochs')
plt.ylabel('Test MSE')
plt.grid(True,linestyle=':',alpha=0.6)
plt.ylim(min(dl_mse_history)*0.8,test_mse_mod*1.5)
plt.legend(facecolor='white', framealpha=1, fontsize=15, edgecolor='black')
plt.show()

print("\n[Final MSE Summary]")
print(f"1. Module : [ Train MSE: {train_mse_mod:.6f}, Test MSE: {test_mse_mod:.6f} ]")
print(f"2. Analytical : [ Train MSE: {train_mse_anal:.6f}, Test MSE: {test_mse_anal:.6f} ]")
print(f"3. Deep Learning : [ Train MSE: {final_train_mse:.6f}, Test MSE: {final_test_mse:.6f} ]")
```
